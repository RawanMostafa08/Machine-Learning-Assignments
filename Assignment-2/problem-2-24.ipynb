### 2.24 (B)

```python
def target_function(x):
    # TODO: implement target function
    return x**2
```

```python
def run_numerical_experiment():
    # TODO: implement experiment
    x_vals = np.linspace(-1, 1, 1000)
    f_x = target_function(x_vals)
    h = []

    for _ in range(100000):
        x1, x2 = np.random.uniform(-1, 1, 2)
        a = x1 + x2
        b = -x1 * x2
        h_x = a * x_vals + b
        h.append(h_x)

    h = np.array(h)
    g_x = np.mean(h, axis=0)
    bias = (g_x - f_x)**2
    var_x = np.mean((h - g_x)**2, axis=0)
    e_out = np.mean((h - f_x)**2)

    print(f"E_out: {e_out}")
    print(f"Bias(x) (avg): {np.mean(bias)}")
    print(f"Variance (avg): {np.mean(var_x)}")
    print(f"Bias + Variance : {np.mean(bias) + np.mean(var_x)}")

    plt.plot(x_vals, f_x, label='f(x) = x²', color='blue')
    plt.plot(x_vals, g_x, label='g(x)', color='red')
    plt.legend()
    plt.grid()
    plt.show()
```

### 2.24 (E)

```python
# TODO: Run experiment for quadratic hypothesis

def run_numerical_experiment_constant(n):
    x_vals = np.linspace(-1, 1, 1000)
    f_x = target_function(x_vals)
    h_values = []

    for _ in range(100000):
        x_train = np.random.uniform(-1, 1, n)
        y_train = x_train ** 2
        c = np.mean(y_train) 
        h_x = np.full_like(x_vals, c)
        h_values.append(h_x)

    h_values = np.array(h_values)
    g_x = np.mean(h_values, axis=0)
    bias = (g_x - f_x) ** 2
    var_x = np.mean((h_values - g_x) ** 2, axis=0)
    e_out = np.mean((h_values - f_x) ** 2)

    print(f"E_out: {e_out}")
    print(f"Bias(x) (avg): {np.mean(bias)}")
    print(f"Variance (avg): {np.mean(var_x)}")
    print(f"Bias + Variance : {np.mean(bias) + np.mean(var_x)}")

    plt.plot(x_vals, f_x, label='f(x) = x²', color='blue')
    plt.plot(x_vals, g_x, label='g(x)', color='red')
    plt.title(f"n = {n}")
    plt.legend()
    plt.grid()
    plt.show()


run_numerical_experiment_constant(2)
run_numerical_experiment_constant(5)
run_numerical_experiment_constant(10)
run_numerical_experiment_constant(20)
```

```python
# variance decreases as number of training points increases
# bias almost doesn't change as number of training points increases
# Bias + variance approximately equals E_out
# Eout decreases as number of training points increases
# bias > variance because we're using a simple model
```
